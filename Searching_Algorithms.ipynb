```python
# Linear Search- One by One
def linear_search(numbers, target):

    # Check every element
    for i in range(len(numbers)):

        if numbers[i] == target:
            return i
                # Target not found
    return -1
```

```python
numbers = [10, 25, 30, 45, 60]
```

```python
print(linear_search(numbers, 45))
```

```python
print(linear_search(numbers, 99))
```

**Linear Search with Comparison Counter**

```python
def linear_search(numbers, target):

    comparisons = 0

    for i in range(len(numbers)):

        comparisons += 1

        if numbers[i] == target:
            print("Found!")
            print("Comparisons:", comparisons)
            return i

    print("Not Found")
    print("Comparisons:", comparisons)

    return -1
```

```python
numbers = [10, 20, 30, 40, 50, 60, 70, 60, 60]
```

```python
linear_search(numbers, 560)
```

**Binary Search**

```python
numbers = [10, 20, 30, 40, 50, 60, 70]
```

```python
len(numbers)
```

```python
target = 50
low = 0
high = len(numbers) - 1

while low <= high:

    mid = (low + high) // 2

    if numbers[mid] == target:
        print("Found at index:", mid)
        break

    elif numbers[mid] < target:
        low = mid + 1

    else:
        high = mid - 1
```
