Q.1 A sales manager has monthly sales figure for 5 people. Create a synthetic list of sales figures and identify who achieved the target and calculate the number of people who achieved it.

```python
# My Response
sales = [50000, 70000, 45000, 80000, 60000]

target = 60000

for sale in sales:
    if sale >= target:
        print(sale, "Achieved")
    else:
        print(sale, "Not Achieved")

print("People achieved:", sum(sale >= target for sale in sales))
```

Q.2 A customer wants to purchase multiple products. You are supposed to ask how many products they want to buy and then collect product name,quantity and price
and calculate bill also

```python
#My response
def buy_products():

    n = int(input("Enter number of products:"))
    total = 0

    for i in range(n):
        name = input("Enter product name:")
        quantity = int(input("Enter quantity:"))
        price = float(input("Enter price:"))

        bill = quantity * price
        total = total + bill

        print(name, "Bill =", bill)

    print("Total Bill =", total)

buy_products()
```

```python
#Teacher response
def calculate_bill():
  cart=[]

  n=int(input("How many products do you want to buy?"))

  for i in range(n):
    name=input("Enter product name:")
    quantity=int(input("Enter quantity:"))
    price=float(input("Enter price:"))

    total_price=quantity*price
    cart.append

  print("Total Bill:",total_price)

calculate_bill()
```
