```python
text = ""
undo_stack = []

def type_text(new_text):
    global text

    # Save current text before making a change
    undo_stack.append(text)

    # Add new text
    text = text + new_text

    print("Text:", text)


def undo():
    global text

    if undo_stack:
        text = undo_stack.pop()
        print("Undo:", text)
    else:
        print("Nothing to undo")


# Try the editor
type_text("Hello ")
type_text("World")
type_text("!")

print("\nPerforming Undo operations:")
undo()
undo()
undo()
undo()
```

```python
# ============================================================
# ACTIVITY 5: BROWSER BACK BUTTON
# Stack example
# ============================================================

history = []
history.append("Google")
history.append("YouTube")
history.append("Instagram")
history.append("LinkedIn")

print("\nBrowser History:", history)
print("Current page:", history[-1])

# Back button
history.pop()

print("After Back:", history)
print("Current page:", history[-1])
```

```python
# ============================================================
# ACTIVITY 6: BASIC QUEUE
# FIFO = First In, First Out
# ============================================================

from collections import deque
queue = deque()
# Add customers
queue.append("A")
queue.append("B")
queue.append("C")

print("\nQueue:", queue)
# First customer
print("Front:", queue[0])
# Remove first customer
removed = queue.popleft()

print("Removed:", removed)
print("Queue after removal:", queue)
```

```python
# ============================================================
# ACTIVITY 7: CUSTOMER SERVICE QUEUE
# ============================================================
from collections import deque
customers = deque()
customers.append("Rahul")
customers.append("Priya")
customers.append("Amit")
customers.append("Sneha")

print("\nWaiting customers:", customers)
while customers:
    customer = customers.popleft()
    print("Serving:", customer)
```

```python
# ============================================================
# ACTIVITY 8: QUEUE WITH NUMBERS
# ============================================================

from collections import deque

queue = deque()
queue.append(10)
queue.append(20)
queue.append(30)
queue.append(40)

print("\nQueue:", queue)
```

```python
print("First customer:", queue[0])
print("Served:", queue.popleft())
print("Remaining:", queue)
```

```python
# ============================================================
# ACTIVITY 9: BANK EXAMPLE
# Queue = Customers
# Stack = Transactions
# ============================================================
from collections import deque

# Customer Queue
customers = deque()

customers.append("Rahul")
customers.append("Priya")
customers.append("Amit")

# Transaction Stack
transactions = []

transactions.append("Deposit ₹5000")
transactions.append("Withdraw ₹1000")
transactions.append("Deposit ₹2000")

print("\nCustomers:", customers)
print("Transactions:", transactions)
```

```python
# Serve first customer
print("\nServing:", customers.popleft())

# Undo latest transaction
print("Undo:", transactions.pop())

print("\nRemaining customers:", customers)
print("Remaining transactions:", transactions)
```

```python
# ============================================================
# ACTIVITY 10: SIMPLE STACK vs QUEUE DEMO
# ============================================================
# STACK
stack = []

stack.append("A")
stack.append("B")
stack.append("C")

print("\nSTACK")
print("Before:", stack)
print("Removed:", stack.pop())
print("After:", stack)

# QUEUE
queue = deque()

queue.append("A")
queue.append("B")
queue.append("C")

print("\nQUEUE")
print("Before:", queue)
print("Removed:", queue.popleft())
print("After:", queue)
```
