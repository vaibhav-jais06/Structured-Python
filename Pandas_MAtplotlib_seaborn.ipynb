```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Set a consistent visual style
sns.set_theme(style="whitegrid")
print("Libraries loaded successfully!")
```

```python
# SECTION 1 — PANDAS BASICS
# Goal: Create, inspect, filter, sort, and summarize a small dataset
# ================================================================

# Step 1: Create a small student marks dataset
student_data = {
    "Name": ["Asha", "Rahul", "Meera", "Arjun", "Sara", "Dev"],
    "Subject": ["Math", "Science", "Math", "Science", "Math", "Science"],
    "Marks": [85, 72, 91, 44, 67, 58],
    "Study_Hours": [5, 3, 6, 1, 4, 2]
}
```

```python
student_data
```

```python
df = pd.DataFrame(student_data)
df.head()
```

```python
print("First five rows:")
df.head()
```

```python
print("\nDataset shape (rows, columns):")
print(df.shape)
```

```python
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Step 3: Check data types and summary statistics
print("\nData types:")
print(df.dtypes)

print("\nSummary of numeric columns:")
df.describe()
```

```python
# Step 4: Select columns
print("\nOnly Name and Marks:")
print(df[["Name", "Marks"]])
```

```python
# Step 5: Filter rows — students scoring at least 70
high_scorers = df[df["Marks"] >= 70]
print("\nStudents scoring 70 or above:")
print(high_scorers)
```

```python
# Step 6: Sort from highest to lowest marks
sorted_df = df.sort_values("Marks", ascending=True)
print(sorted_df)
```

```python
# Step 7: Create a new column
df["Result"] = np.where(df["Marks"] >= 50, "Pass", "Needs Support")
print("\nAdded Result column:")
print(df)
```

```python
# Step 8: Group and calculate averages
average_by_subject = df.groupby("Subject")["Marks"].mean()
average_by_subject
```

```python
# Step 9: Demonstration of  missing values in the data-np.nan[not a number]
df_missing = df.copy()
df_missing.loc[2, "Study_Hours"] = np.nan
df_missing
```

```python
print("\nMissing values in each column:")
print(df_missing.isnull().sum())
```

```python
# Fill the missing study-hours value with the column average
df_missing["Study_Hours"] = df_missing["Study_Hours"].fillna(df_missing["Study_Hours"].mean()
)
print(df_missing)
```

# **Data visualization using Matplotlib**

```python
# SECTION 2 — MATPLOTLIB BAR CHART
# Goal: Compare categories
# ================================================================
batsmen = ["Kohli", "Rohit", "Gill", "Dhoni", "Pandya", "Iyer"]
runs = [85, 72, 91, 44, 67, 58]
plt.figure(figsize=(8, 4))
plt.bar(batsmen, runs, edgecolor="black")
plt.title("IPL Batsmen — Runs Scored")
plt.xlabel("Batsman")
plt.ylabel("Runs")
```

```python
# ================================================================
# SECTION 3 — MATPLOTLIB LINE CHART
# Goal: Show change across matches
# ================================================================
matches = [1, 2, 3, 4, 5]
kohli_runs = [45, 72, 38, 91, 55]
rohit_runs = [60, 33, 88, 42, 70]
plt.figure(figsize=(8, 4))
plt.plot(matches, kohli_runs, marker="o", label="Kohli")
plt.plot(matches, rohit_runs, marker="s", label="Rohit")
plt.title("Runs Across Five Matches")
plt.xlabel("Match Number")
plt.ylabel("Runs")
plt.legend()
```

```python
# SECTION 4 — MATPLOTLIB SCATTER PLOT
# Goal: Explore a relationship between two numeric variables
# ================================================================
strike_rate = [142, 138, 155, 110, 148, 130]
plt.figure(figsize=(7, 4))
plt.scatter(runs, strike_rate, s=100, alpha=0.8)
# Add each batsman's name beside the point
for i, name in enumerate(batsmen):
  plt.annotate(name, (runs[i], strike_rate[i]), xytext=(5, 4),
                 textcoords="offset points")
plt.title("Runs vs Strike Rate")
plt.xlabel("Runs")
plt.ylabel("Strike Rate")
```

```python
# ================================================================
# SECTION 5 — PIE CHART + HISTOGRAM
# Goal: Parts of a whole and distribution
# ================================================================
# Pie chart: team win counts
teams = ["MI", "CSK", "RCB", "KKR", "DC", "Others"]
wins = [5, 5, 2, 3, 2, 3]
plt.figure(figsize=(6, 6))
plt.pie(wins, labels=teams, autopct="%1.1f%%", startangle=90)
plt.title("Share of Wins by Team")
plt.show()
```

```python
# Histogram: create reproducible sample student marks
np.random.seed(10)
marks = np.random.normal(loc=68, scale=12, size=200).clip(0, 100)
plt.figure(figsize=(8, 4))
plt.hist(marks, bins=15, edgecolor="white",color='yellow')
plt.axvline(marks.mean(), linestyle="--",
            label=f"Average = {marks.mean():.1f}")
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```
