# Object-Oriented Programming System (OOPS)

### Advantages of OOP:
- **Easy to understand:** Modular design reflecting real-world concepts.
- **Safe & Secure:** Data abstraction and encapsulation protect sensitive states.
- **Easy to Fix & Maintain:** Isolated classes and methods streamline debugging.

### Four Pillars of OOP:
1. **Encapsulation:** Bundling data and methods; managing access via public, protected (`_`), and private (`__`) members.
2. **Abstraction:** Hiding complex implementation details and exposing clean interfaces.
3. **Inheritance:** Deriving child classes from parent classes for code reusability (`super()`).
4. **Polymorphism:** Unified interfaces allowing different classes to implement methods differently.

### Key Definitions:
- **Class:** Blueprint / template for creating objects.
- **Object:** Concrete instance of a class.


```python
#NormalWay (Without OOPS)
name="Priya"
marks=90
print(name,marks)
```

```python
#Design-One Design Many objects
class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
s1=student("Priya",90)
s2=student("Rahul",85)
s3=student("Rohit",90)
```

```python
#2. METHODS AND CLASS VARIABLES
# Class variables are shared by all objects.
print("/n2. EMPLOYEE CLASS")

class Employee:

  count=0

  def __init__(self,name,salary):
    self.name=name
    self.salary=salary
    Employee.count+=1

  def display(self):
    print("Employee:",self.name)
    print("salary:",self.salary)

e1=Employee("John",25000)
e2=Employee("Alice",30000)
e3=Employee("Rahul",35000)

e1.display()
e2.display()
e3.display()

print("Total Employee:",Employee.count)
```

```python
#Encapsulation
print("/n3. ENCAPSULATION")

class BankAccount:

  def __init__(self,name,balance):
    self.name=name   #Public
    self._branch="Bangalore"   #Protected
    self.__balance= balance  #Private

  def show_balance(self):
    #Private variable accesses inside the class
    print("Account balance:", self.__balance)
```

```python
account=BankAccount("Amith",5000)
print("Name:",account.name)
print("Branch:",account._branch)
account.show_balance()
```

```python
# 4. INHERITANCE
#Child class inherits the properties of the parent class
print("/n3. INHERITANCE")

class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age

  def introduce(self):
    print("My name is",self.name)
    print("My age is",self.age)

class StudentDetails(Person):
  def __init__(self,name,age,roll_no):
    super().__init__(name,age)
    self.roll_no=roll_no

  def display(self):
      self.introduce()
      print("Roll Number:",self.roll_no)

student=StudentDetails("Aman",20,101)
student.display()
```

```python
#5. Polymorphism
print("/n3. POLYMORPHISM")

class Dog:
  def sound(self):
    print("Dog barks")
class Cat:
  def sound(self):
    print("Cat meows")
class Cow:
  def sound(self):
    print("Cow moos")

#Create Objects
animals=[Dog(),Cat(),Cow()]

#Same method called for different objects
for animal in animals:
  animal.sound()
```

```python
#Method Overriding
print("/n3. METHOD OVERRIDING")

class Animal:
  def sound(self):
    print("Animal makes a sound")

class Lion(Animal):
  def sound(self):
    print("Lion roars")

class Elephant(Animal):
  def sound(self):
    print("Elephant trumpets")

a1=Animal()
a2=Lion()
a3=Elephant()

a1.sound()
a2.sound()
a3.sound()
```
